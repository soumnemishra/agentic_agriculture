```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

def generate_paper_plots():
    # Load the fixed dataset
    csv_path = r'D:\agentic_agriculture\datasets\experiment_results.csv'
    if not os.path.exists(csv_path):
        print(f"Error: Could not find {csv_path}")
        return
        
    df = pd.read_csv(csv_path)
    
    # Set academic formatting
    plt.style.use('seaborn-v0_8-paper')
    sns.set_context("paper", font_scale=1.2)
    
    # --- PLOT 1: Epistemic Belief Revision (Sensor Fusion) ---
    # This proves that your Reasoning Layer is actually thinking, 
    # not just blindly trusting the Vision model.
    plt.figure(figsize=(9, 6))
    sns.scatterplot(
        data=df, 
        x='vision_confidence', 
        y='reasoning_posterior', 
        hue='vision_predicted_class',
        style='vision_predicted_class', 
        s=120, 
        alpha=0.8
    )
    plt.plot([0, 1], [0, 1], 'k--', label='No Fusion Effect ($y=x$)')
    plt.title('Impact of Microclimate Sensor Fusion on Diagnostic Confidence')
    plt.xlabel('Vision Model Confidence (Optical Prior)')
    plt.ylabel('Reasoning Agent Confidence (Posterior Belief)')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('datasets/fig1_sensor_fusion.png', dpi=300)
    
    # --- PLOT 2: Autonomous Intervention Timeline ---
    # This proves the system takes appropriate physical actions 
    # as the farm state changes over time.
    plt.figure(figsize=(12, 5))
    sns.scatterplot(
        data=df, 
        x='step_index', 
        y='planning_action', 
        hue='vision_predicted_class', 
        s=120,
        palette='Dark2'
    )
    plt.title('Autonomous Agronomic Interventions Over Time')
    plt.xlabel('Simulation Step (18-second intervals)')
    plt.ylabel('Executed Actuator Command')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('datasets/fig2_action_timeline.png', dpi=300)

    print("Success: High-resolution plots saved to datasets/ directory.")

if __name__ == "__main__":
    generate_paper_plots()
```
